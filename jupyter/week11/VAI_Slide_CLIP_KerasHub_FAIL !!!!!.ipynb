```python
import keras
import numpy as np
import tensorflow as tf
from keras_hub.models import CLIPBackbone, CLIPPreprocessor
```

### **Load the Model and Preprocessor**

```python
# Choose a preset, e.g., 'clip_vit_base_patch32'
PRESET = "clip_vit_base_patch32"

# 1. Load the pre-trained CLIP model (the dual encoder)
# Load the model architecture without pre-trained weights to avoid the loading error.
#clip_model = CLIPBackbone.from_preset(PRESET, load_weights=false)

HUGGINGFACE_PRESET = "hf://keras/clip_vit_base_patch32"
clip_model = CLIPBackbone.from_preset(HUGGINGFACE_PRESET)


# 2. Load the preprocessor which handles image and text formatting
preprocessor = CLIPPreprocessor.from_preset(PRESET)
```

### **Prepare Inputs (Image and Text Labels)**

CLIP works by comparing image embeddings to text embeddingss (work as labels Y).

Image: Load and convert an image to a NumPy array.
Text: Create a list of candidate labels/prompts.

```python
# --- 1. Image Data ---
# Create a dummy image (e.g., a 224x224 RGB image) for demonstration.
# Replace this with your actual image loading code.
image_data = np.random.randint(0, 256, size=(224, 224, 3), dtype=np.uint8)
image_batch = np.expand_dims(image_data, axis=0) # Add batch dimension

# --- 2. Text Labels (Prompts) ---
candidate_prompts = [
    "a photo of a cat",
    "a picture of a dog",
    "a portrait of a car",
    "a photograph of a bird"
]
```

### **Preprocess and Encode Inputs**

Pass the raw image and text to the **CLIPPreprocessor**. This layer will convert them into the token IDs and tensors that the **CLIPBackbone** expects.

Variable **"preprocessor"** is from **"CLIPPreprocessor"** in **KerasHub**

Variable **"clip_model"** is from **"CLIPBackbone"** in **KerasHub**

```python
# Preprocess the data,
inputs = preprocessor(
    {"images": image_batch, "prompts": candidate_prompts}
)

# Encode the inputs to get image and text features (embeddings)
outputs = clip_model(inputs)

# The outputs contain normalized features/embeddings
# Corrected keys based on the model's output structure
image_features = outputs["vision_logits"]
text_features = outputs["text_logits"]
```

### **Compute Cosine Similarity as performing Zero-Shot Prediction**

To perform zero-shot classification, calculate the cosine similarity between the **ONE** image embedding and **ALL** the text embeddings. The text with the highest similarity score is the predicted class / text.

```python
# Calculate the cosine similarity between the image and all text features
# This is equivalent to a dot product because the features are already normalized
similarity_scores = tf.linalg.matmul(
    image_features, text_features, transpose_b=True
)
# similarity_scores is a (1, N) tensor, where N is the number of prompts

# Convert scores to probabilities using softmax
probabilities = tf.nn.softmax(similarity_scores / clip_model.temperature.numpy(), axis=-1)

# Get the predicted index
predicted_index = tf.argmax(probabilities, axis=-1).numpy()[0]
predicted_label = candidate_prompts[predicted_index]

# --- Print Results ---
print(f"Candidate Prompts: {candidate_prompts}")
print(f"Similarity Scores (Logits): {similarity_scores.numpy()}")
print(f"Probabilities: {probabilities.numpy()}")
print(f"\nPredicted Class: {predicted_label}")
```
