```python
import numpy as np
import tensorflow as tf
from PIL import Image
import matplotlib.pyplot as plt
```

```python
def preprocess(image_path, mask_path):
    # Load and normalize image
    img = tf.io.read_file(image_path)
    img = tf.image.decode_png(img, channels=3)
    img = tf.image.resize(img, [256, 256])
    img = tf.cast(img, tf.float32) / 255.0

    # Load and normalize mask (assuming binary mask)
    mask = tf.io.read_file(mask_path)
    mask = tf.image.decode_png(mask, channels=1)
    mask = tf.image.resize(mask, [256, 256], method=tf.image.ResizeMethod.NEAREST_NEIGHBOR)
    mask = tf.cast(mask, tf.float32) / 255.0

    return img, mask
```

```python
train_img_paths = ["Dog.png"]
train_mask_paths = ["Dog_SAM_Labels.png"]
```

```python
img_input, mask_true = preprocess(train_img_paths[0], train_mask_paths[0])

Tot_Class = len(np.unique(mask_true))
print(f"Total number of classes found: {Tot_Class}")
```

```python
# Create dataset

train_dataset = tf.data.Dataset.from_tensor_slices((train_img_paths, train_mask_paths))

BATCH_SIZE = 1
AUTOTUNE = tf.data.AUTOTUNE # Define AUTOTUNE

train_dataset = train_dataset.map(preprocess, num_parallel_calls=AUTOTUNE)
train_dataset = train_dataset.batch(BATCH_SIZE)
train_dataset = train_dataset.prefetch(AUTOTUNE)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

def conv_block(inputs, num_filters):
    # Standard Double Convolution Block
    x = layers.Conv2D(num_filters, 3, padding="same")(inputs)
    x = layers.BatchNormalization()(x)
    x = layers.Activation("relu")(x)

    x = layers.Conv2D(num_filters, 3, padding="same")(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation("relu")(x)
    return x

def encoder_block(inputs, num_filters):
    x = conv_block(inputs, num_filters)
    p = layers.MaxPool2D((2, 2))(x)
    return x, p

def decoder_block(inputs, skip_features, num_filters):
    # Upsample
    x = layers.Conv2DTranspose(num_filters, (2, 2), strides=2, padding="same")(inputs)
    # Copy and Crop (Skip Connection)
    x = layers.Concatenate()([x, skip_features])
    # Conv Block
    x = conv_block(x, num_filters)
    return x

def build_unet(input_shape, num_classes):
    inputs = layers.Input(input_shape)

    # --- Encoder (Downsampling) ---
    s1, p1 = encoder_block(inputs, 64)    # Output: 128x128 (if input 256)
    s2, p2 = encoder_block(p1, 128)       # Output: 64x64
    s3, p3 = encoder_block(p2, 256)       # Output: 32x32
    s4, p4 = encoder_block(p3, 512)       # Output: 16x16

    # --- Bridge (Bottleneck) ---
    b1 = conv_block(p4, 1024)             # Output: 16x16x1024

    # --- Decoder (Upsampling) ---
    d1 = decoder_block(b1, s4, 512)       # In: 16x16 + 16x16 -> Out: 32x32
    d2 = decoder_block(d1, s3, 256)       # In: 32x32 + 32x32 -> Out: 64x64
    d3 = decoder_block(d2, s2, 128)       # In: 64x64 + 64x64 -> Out: 128x128
    d4 = decoder_block(d3, s1, 64)        # In: 128x128 + 128x128 -> Out: 256x256

    # --- Output ---
    if num_classes == 1:
        activation = "sigmoid" # Binary segmentation
    else:
        activation = "softmax" # Multi-class segmentation

    outputs = layers.Conv2D(num_classes, 1, padding="same", activation=activation)(d4)

    model = models.Model(inputs, outputs, name="U-Net")
    return model

# Initialize the model
model = build_unet((256, 256, 3), num_classes=Tot_Class)
```

```python
model.summary()
```

```python
import tensorflow.keras.backend as K

def dice_coef(y_true, y_pred, smooth=100):
    y_true_f = K.flatten(y_true)
    y_pred_f = K.flatten(y_pred)
    intersection = K.sum(y_true_f * y_pred_f)
    dice = (2. * intersection + smooth) / (K.sum(y_true_f) + K.sum(y_pred_f) + smooth)
    return dice

def dice_loss(y_true, y_pred):
    return 1 - dice_coef(y_true, y_pred)
```

```python
from tensorflow.keras.callbacks import ModelCheckpoint, ReduceLROnPlateau, EarlyStopping

# Compile
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
    loss=dice_loss, # Or "binary_crossentropy"
    metrics=[dice_coef, 'accuracy']
)

"""
# Callbacks
callbacks = [
    ModelCheckpoint("best_unet_model.keras", save_best_only=True),
    ReduceLROnPlateau(monitor='val_loss', factor=0.1, patience=5, min_lr=1e-7),
    EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
]
"""

# Train

# Define a mapping function to one-hot encode the mask
def one_hot_encode_mask(img, mask):
    mask = tf.squeeze(mask, axis=-1) # Remove the last dimension (channel=1)
    mask = tf.cast(mask * 255, tf.int32) # Convert normalized float back to integer class IDs
    mask = tf.one_hot(mask, depth=Tot_Class) # One-hot encode
    return img, mask

# Apply the one-hot encoding to the dataset
train_dataset_one_hot = train_dataset.map(one_hot_encode_mask, num_parallel_calls=AUTOTUNE)
```

```python
history = model.fit(
    train_dataset_one_hot,
#    validation_data=val_dataset,
    epochs=2500,
#    callbacks=callbacks
)
```

```python
# --- 4. PREDICTION ---
prediction = model.predict(train_dataset) # Output will be (1, 388, 388, 1)
pred_mask = (prediction > 0.5).astype(np.float32)
```

```python
# --- 5. COMPUTE IOU & DICE ---
def compute_metrics(y_true, y_pred):
    # Resize true mask to match the smaller output of 'valid' padding U-Net
    y_true_resized = tf.image.resize(y_true, (y_pred.shape[1], y_pred.shape[2]), method="nearest").numpy()

    intersection = np.sum(y_true_resized * y_pred)
    union = np.sum(y_true_resized) + np.sum(y_pred) - intersection

    iou = intersection / (union + 1e-7)
    dice = (2. * intersection) / (np.sum(y_true_resized) + np.sum(y_pred) + 1e-7)

    return iou, dice, y_true_resized
```

```python
iou, dice, true_resized = compute_metrics(mask_true, pred_mask)
print(f"IoU: {iou:.4f}, Dice: {dice:.4f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ---- convert GT / Pred to label maps (H,W) ----
gt = true_resized[:, :, 0]
pr = pred_mask[0]

# If multi-class probs/logits: (H,W,C) -> argmax labels
if pr.ndim == 3 and pr.shape[-1] > 1:
    pr_lab = np.argmax(pr, axis=-1).astype(np.int32)
else:
    # binary or already-label mask: (H,W,1) or (H,W)
    pr_lab = (pr[..., 0] if pr.ndim == 3 else pr).astype(np.int32)

# If GT is one-hot: (H,W,C) -> argmax labels, else assume already labels in (H,W)
if gt.ndim == 3 and gt.shape[-1] > 1:
    gt_lab = np.argmax(gt, axis=-1).astype(np.int32)
else:
    gt_lab = gt.astype(np.int32)

# ---- choose a discrete colormap (each class a color) ----
vmax = int(max(gt_lab.max(), pr_lab.max()))
cmap = plt.get_cmap("tab20", vmax + 1)  # good default for up to ~20 classes

# ---- 6. VISUALIZATION (3 panels) ----
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.title("Raw Image (Input)")
plt.imshow(img_input)
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title("Ground Truth (Colored)")
plt.imshow(gt_lab, cmap=cmap, vmin=0, vmax=vmax)
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title(f"Prediction (Colored)\nIoU: {iou:.2f}")
plt.imshow(pr_lab, cmap=cmap, vmin=0, vmax=vmax)
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# --- SETTINGS ---
ALPHA = 0.5  # Transparency of the mask
THRESHOLD = 0.5

# 1. Prepare the mask data (Binarize the prediction)
# If pred_mask is (1, 256, 256, 1), we squeeze it to (256, 256)
p_mask = (pred_mask[0, :, :, 0] > THRESHOLD).astype(np.float32)
t_mask = true_resized[:, :, 0]

# 2. Create Colored Masks (using RGB)
# Ground Truth = Green [0, 1, 0]
gt_color = np.zeros((*t_mask.shape, 3))
gt_color[..., 1] = t_mask

# Prediction = Red [1, 0, 0] (or choose Blue [0, 0, 1])
pred_color = np.zeros((*p_mask.shape, 3))
pred_color[..., 0] = p_mask

# --- 6. VISUALIZATION ---
plt.figure(figsize=(18, 6))

# Subplot 1: Raw Image
plt.subplot(1, 3, 1)
plt.title("Raw Image (Input)")
plt.imshow(img_input)
plt.axis('off')

# Subplot 2: Ground Truth Overlay (Green)
plt.subplot(1, 3, 2)
plt.title("Ground Truth Overlay")
plt.imshow(img_input)
plt.imshow(gt_color, alpha=ALPHA) # Overlay green mask
plt.axis('off')

# Subplot 3: Prediction Overlay (Red)
plt.subplot(1, 3, 3)
plt.title(f"Prediction Overlay\nIoU: {iou:.4f} | Dice: {dice:.4f}")
plt.imshow(img_input)
plt.imshow(pred_color, alpha=ALPHA) # Overlay red mask
plt.axis('off')

plt.tight_layout()
plt.show()
```
