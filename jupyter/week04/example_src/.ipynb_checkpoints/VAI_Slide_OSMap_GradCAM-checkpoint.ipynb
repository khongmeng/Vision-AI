```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from google.colab import files
from tensorflow.keras.applications.inception_v3 import InceptionV3, preprocess_input, decode_predictions
from tensorflow.keras.preprocessing import image
```

```python
# 1. Load GoogLeNet equivalent (InceptionV3)
model = InceptionV3(weights='imagenet')
```

```python
uploaded = files.upload()
filename = list(uploaded.keys())[0]

img = keras.utils.load_img(filename, target_size=(299, 299))
```

```python
# 2. Load and Resize image (InputSize for InceptionV3 is 299x299)
x = keras.utils.img_to_array(img)
x = np.expand_dims(x, axis=0)

print(x.shape)
x = np.expand_dims(x, axis=0)              # shape: (1, 299, 299, 3)
print(x.shape)
```

```python
x_preprocessed = preprocess_input(x.copy())
```

```python
# 3. Classify (Get Label)
preds = model.predict(x_preprocessed[0])
label_index = np.argmax(preds[0])
print('Predicted:', decode_predictions(preds, top=1)[0])

# 4. Occlusion Sensitivity (Manual Implementation)
def get_occlusion_map(model, img_array, label_idx, patch_size=64):
    height, width, _ = img_array.shape
    score_map = np.zeros((height, width))

    # Slide a "gray" patch across the image
    for i in range(0, height, patch_size):
        for j in range(0, width, patch_size):
            occluded_img = img_array.copy()
            # Apply gray patch (occlusion)
            occluded_img[i:i+patch_size, j:j+patch_size, :] = 128

            # Predict on occluded image
            out = model.predict(np.expand_dims(preprocess_input(occluded_img), axis=0), verbose=0)
            # Score is the probability of the original class
            score_map[i:i+patch_size, j:j+patch_size] = out[0][label_idx]

    # Sensitivity = 1 - score (high value = high importance)
    return 1 - score_map

score_map = get_occlusion_map(model, x[0][0], label_index)

# 5. Visualization (Imshow + Alpha Overlay)
plt.imshow(img)
plt.imshow(score_map, cmap='jet', alpha=0.5)
plt.colorbar()
plt.show()
```

```python
from tensorflow.keras.applications.inception_v3 import InceptionV3, preprocess_input
from tensorflow.keras.models import Model
```

```python
# For Grad-CAM, we need the output of the LAST conv layer
# In InceptionV3, this is usually 'mixed10'
last_conv_layer_name = "mixed10"

# 2. Create a model that maps input to (Last Conv Layer, Predictions)
grad_model = Model(
    [model.inputs], [model.get_layer(last_conv_layer_name).output, model.output]
)
```

```python
def compute_gradcam(img_array, class_index):
    with tf.GradientTape() as tape:
        last_conv_layer_output, preds = grad_model(img_array)
        loss = preds[:, class_index]

    # Gradient of the class score with respect to the feature map
    grads = tape.gradient(loss, last_conv_layer_output)

    # Mean intensity of the gradient over each feature map channel
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))

    # Multiply each channel by "how important it is"
    last_conv_layer_output = last_conv_layer_output[0]
    heatmap = last_conv_layer_output @ pooled_grads[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)

    # Normalize heatmap between 0 and 1
    heatmap = tf.maximum(heatmap, 0) / tf.math.reduce_max(heatmap)
    return heatmap.numpy()
```

```python
# 3. Classify (Get Label)
img = x_preprocessed[0]
preds = model.predict(img)
print('Predicted:', decode_predictions(preds, top=1)[0])
```

```python
pred_index = np.argmax(preds[0])
heatmap = compute_gradcam(img, pred_index)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing import image

# The 'img' variable in this cell refers to the preprocessed image (x_preprocessed[0]).
# For visualization, we need the original image (0-255 pixel values, uint8).
# The 'x' variable from cell mh0ZbBcQfIrx contains the original image data before final preprocessing.
# Specifically, x[0][0] holds the (299, 299, 3) un-preprocessed image array (values 0-255, dtype float32).
original_image_for_display = x[0][0].astype(np.uint8)

# 2. Resize heatmap to the original image size
# original_image_for_display.shape[1] is width, original_image_for_display.shape[0] is height
heatmap_resized = cv2.resize(heatmap, (original_image_for_display.shape[1], original_image_for_display.shape[0]))

# 3. Convert heatmap to RGB using the 'JET' colormap (MATLAB style)
heatmap_uint8 = np.uint8(255 * heatmap_resized)
heatmap_color = cv2.applyColorMap(heatmap_uint8, cv2.COLORMAP_JET)
# cv2.applyColorMap typically returns BGR. Convert to RGB for matplotlib.
heatmap_color_rgb = cv2.cvtColor(heatmap_color, cv2.COLOR_BGR2RGB)

# 4. Superimpose the heatmap on original image (Alpha = 0.5)
# Both inputs to cv2.addWeighted must have the same shape and data type (e.g., uint8).
superimposed_img = cv2.addWeighted(original_image_for_display, 0.6, heatmap_color_rgb, 0.4, 0)

# 5. Visual Output
plt.figure(figsize=(10, 10))
plt.imshow(superimposed_img)
plt.axis('off')
plt.title("Grad-CAM: What the model is 'looking' at")
plt.show()
```
