# **Create Backbone of Autoencoder (AE)**

```python
def create_autoencoder(input_dim, encoding_dim=32):
    input_layer = layers.Input(shape=(input_dim,))
    x = layers.Dense(input_dim // 2, activation='relu')(input_layer)  # 48 units
    x = layers.Dense(input_dim // 4, activation='relu')(x)            # 24 units
    encoded = layers.Dense(encoding_dim, activation='relu')(x)        # 32 units

    x = layers.Dense(input_dim // 4, activation='relu')(encoded)      # 24 units
    x = layers.Dense(input_dim // 2, activation='relu')(x)            # 48 units
    decoded = layers.Dense(input_dim, activation='linear')(x)         # 96 units

    autoencoder = models.Model(input_layer, decoded)
    autoencoder.compile(optimizer='adam', loss='mse')
    return autoencoder
```
